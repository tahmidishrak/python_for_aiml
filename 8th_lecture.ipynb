## Attributes and Methods 

```python
class Car:
    # Class attribute (All cars share this trait)
    wheels = 4 

    def __init__(self, color, brand):
        # Instance attributes (Unique to each car)
        self.color = color
        self.brand = brand

# Creating two separate instances (objects)
car1 = Car("Red", "Toyota")
car2 = Car("Blue", "Ford")

# 1. Reading Attributes
print(car1.brand)   # Output: Toyota (Instance attribute)
print(car2.brand)   # Output: Ford   (Instance attribute)
print(car1.wheels)  # Output: 4      (Shared class attribute)
print(car2.wheels)  # Output: 4      (Shared class attribute)

# 2. Modifying an Instance Attribute
car1.color = "Green"
print(car1.color)   # Output: Green  (Only car1 changes)
print(car2.color)   # Output: Blue   (car2 is completely unaffected)

# 3. Modifying the Class Attribute
Car.wheels = 6
print(car1.wheels)  # Output: 6      (Changes for car1)
print(car2.wheels)  # Output: 6      (Changes for car2 because they share it)
```

```python
class BankAccount:
    # Class Attribute (shared across all instances)
    bank_name = "City Bank"

    def __init__(self, owner, balance):
        # Instance Attributes (unique to each instance)
        self.owner = owner
        self.balance = balance

    # 1. Instance Method
    # Purpose: Operates on data belonging to a specific object.
    # Feature: Takes 'self' as the required first parameter.
    def deposit(self, amount):
        self.balance += amount
        print(f"{self.owner} deposited {amount}. New Balance: {self.balance}")

    # 2. Class Method
    # Purpose: Operates on class-level attributes that affect all instances.
    # Feature: Uses the '@classmethod' decorator and takes 'cls' as the first parameter.
    @classmethod
    def change_bank_name(cls, new_name):
        cls.bank_name = new_name
        print(f"Bank name changed to {cls.bank_name}")

    # 3. Static Method
    # Purpose: Acts as a utility/helper function isolated inside the class.
    # Feature: Uses the '@staticmethod' decorator; does not receive 'self' or 'cls'.
    @staticmethod
    def is_valid_account_number(acc_num):
        # Validates if the account number consists of exactly 10 digits
        return len(acc_num) == 10 and acc_num.isdigit()


# ==================== TESTING THE CODE ====================

# 1. Calling Instance Method
acc1 = BankAccount("Tahmid", 5000)
acc1.deposit(1500)  # Output: Tahmid deposited 1500. New Balance: 6500

# 2. Calling Class Method (directly using the class name)
BankAccount.change_bank_name("DBBL")  # Output: Bank name changed to DBBL

# 3. Calling Static Method (directly using the class name)
print(
    BankAccount.is_valid_account_number("1234567890")
)  # Output: True (10 digits)
```

## Inheritance in OOPs

```python
# Parent Class (Superclass)
class Animal:

    def __init__(self, name):
        self.name = name  # Common attribute for all animals

    def make_sound(self):
        # Generic method to be inherited
        print(f"{self.name} makes a sound.")


# Child Class (Subclass inheriting from Animal)
class Dog(Animal):

    def bark(self):
        # Specific method unique to Dog
        print(f"{self.name} says: Woof Woof!")


# ==================== TESTING THE CODE ====================

# Creating an instance of the Child Class
my_dog = Dog("donky")

# Calling inherited method from Parent Class
my_dog.make_sound()  # Output: Buddy makes a sound.

# Calling specific method from Child Class
my_dog.bark()  # Output: Buddy says: Woof Woof!
```

### Constructor in Inheritance

#### Multilevel Inheritance

```python
# 1. Base Class (Grandparent)
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def start_engine(self):
        return f"The engine of this {self.brand} is now running."

# 2. Intermediate Derived Class (Parent)
class Car(Vehicle):
    def __init__(self, brand, model):
        # Call the constructor of the grandparent class
        super().__init__(brand)
        self.model = model

    def drive(self):
        return f"Driving the {self.brand} {self.model}."

# 3. Sub-Derived Class (Child)
class ElectricCar(Car):
    def __init__(self, brand, model, battery_capacity):
        # Call the constructor of the parent class
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity

    def charge(self):
        return f"Charging the {self.battery_capacity}kWh battery."


# --- Demonstrating the Multilevel Inheritance ---

# Create an object of the bottom-most class (ElectricCar)
my_ev = ElectricCar(brand="Tesla", model="Model 3", battery_capacity=75)

# 1. Access method from the Grandparent class (Vehicle)
print(my_ev.start_engine())  
# Output: The engine of this Tesla is now running.

# 2. Access method from the Parent class (Car)
print(my_ev.drive())         
# Output: Driving the Tesla Model 3.

# 3. Access method from its own Child class (ElectricCar)
print(my_ev.charge())        
# Output: Charging the 75kWh battery.
```

#### Hierarchical Inheritance 

```python
# Base class (Parent)
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_details(self):
        return f"Product: {self.name} | Price: ${self.price}"

# Derived class 1 (Child)
class Electronic(Product):
    def __init__(self, name, price, warranty_months):
        # Initialize the parent class attributes
        super().__init__(name, price)
        self.warranty_months = warranty_months

    def get_electronic_details(self):
        return f"{self.get_details()} | Warranty: {self.warranty_months} months"

# Derived class 2 (Child)
class Clothing(Product):
    def __init__(self, name, price, size):
        # Initialize the parent class attributes
        super().__init__(name, price)
        self.size = size

    def get_clothing_details(self):
        return f"{self.get_details()} | Size: {self.size}"


# --- Demonstration ---

# Create an instance of the first child class
laptop = Electronic("Laptop", 999, 12)
print(laptop.get_electronic_details())
# Output: Product: Laptop | Price: $999 | Warranty: 12 months

# Create an instance of the second child class
shirt = Clothing("T-Shirt", 25, "L")
print(shirt.get_clothing_details())
# Output: Product: T-Shirt | Price: $25 | Size: L
```

#### Hybrid Inharitance 

```python
# Base Class (Hierarchical Root)
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def start_engine(self):
        return f"{self.brand} engine started."

# Child Class 1 (Inherits from Vehicle)
class Car(Vehicle):
    def drive(self):
        return f"{self.brand} is driving on land."

# Child Class 2 (Inherits from Vehicle)
class Boat(Vehicle):
    def sail(self):
        return f"{self.brand} is sailing on water."

# Hybrid Child Class (Combines Multiple and Multilevel Inheritance)
class AmphibiousVehicle(Car, Boat):
    def transform(self):
        return f"{self.brand} is switching modes!"

# --- Driver Code ---
if __name__ == "__main__":
    # Create an instance of the hybrid class
    aquacar = AmphibiousVehicle("PythonCraft")

    # 1. Call method from the ultimate base class (Vehicle)
    print(aquacar.start_engine())  # Output: PythonCraft engine started.

    # 2. Call methods from both intermediate parent classes (Car and Boat)
    print(aquacar.drive())         # Output: PythonCraft is driving on land.
    print(aquacar.sail())          # Output: PythonCraft is sailing on water.

    # 3. Call method from the hybrid class itself
    print(aquacar.transform())     # Output: PythonCraft is switching modes!
```

#### Duck Typing

```python
class Duck:

    def fly(self):
        print("Duck flying high!")


class Airplane:

    def fly(self):
        print("Airplane flying in the sky!")


class Fish:

    def swim(self):
        print("Fish swimming under water.")


# Function that demonstrates Duck Typing
def make_it_fly(entity):
    # Python checks if 'fly()' exists at runtime, regardless of class type
    entity.fly()


# ==================== TESTING ====================

duck = Duck()
plane = Airplane()
fish = Fish()

make_it_fly(duck)  # Output: Duck flying high!
make_it_fly(plane)  # Output: Airplane flying in the sky!

# make_it_fly(fish)  # Raises AttributeError because Fish does not have fly()
```

## Ecapsulation in OOPs

### Type of Encapsulation

#### Public Attributes and Methods

```python
class Car:
    model = "Nissan GTR"
    def model(self):
        print("Nissan GTR SkyLine")
class CopyCar(Car):
    def copyModel(self):
        print(super().model())
obj = CopyCar()
obj.copyModel()
```

#### Protected Attributes and Methods

```python
class Car:
    _model = "Nissan GTR"
    def model(self):
        print("Nissan GTR SkyLine")
class CopyCar(Car):
    def copyModel(self):
        print(super()._model)
obj = CopyCar()
obj.copyModel()
```

#### Privet Attributes and Methods

```python
class Student:

    def __init__(self, name, marks):
        self.name = name  # Public attribute
        self.__marks = marks  # Private attribute (hidden using '__')

    # Public method to view private marks safely
    def get_marks(self):
        return self.__marks


# ==================== TESTING ====================

student1 = Student("Tahmid", 85)

# 1. Accessing public attribute works fine
print(student1.name)  # Output: Tahmid

# 2. Accessing private attribute directly will FAIL/Error
# print(student1.__marks) # Error! Cannot access directly.

# 3. Accessing safely using the public method
print(student1.get_marks())  # Output: 85
```

## Abstraction in OOPs

```python
from abc import ABC, abstractmethod

# Abstract Base Class (Blueprint)
class BankApp(ABC):

    # Concrete Method: Contains sensitive internal database logic
    def dataBase(self):
        print(
            [
                {"Name": "tahmid"},
                {"Age": 15},
                {"Date of Birth": "15 / 02/ 2011"},
                {"pasword": "tahmidishrak4246@"},
            ]
        )

    # Abstract Method: Forces child classes to implement their own security check
    @abstractmethod
    def security(self):
        pass


# Child Class inheriting from BankApp
class Data_1(BankApp):

    # Custom method to access database via parent class
    def details(self):
        print(super().dataBase())

    # Implementing the compulsory abstract method from parent
    def security(self):
        print("Successfully connected with security")
        
# ==================== CODE EXECUTION ====================
# Object creation of child class
obj = Data_1()
# Calling the details method
obj.details()
```
