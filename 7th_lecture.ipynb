# Exception Handling 

## Exception Handling Keyword 

### Try Keyword 

```python
a = int(input("Give your number: "))
try:
    print(10/a)
except Exception as err:
    print(f"There is an error as {err}")
else:
    print("Good no occurs have this code")
finally:
    print("i will run what matter")
print("Complete")
```
