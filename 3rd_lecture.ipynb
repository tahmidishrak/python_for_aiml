# If-Else and Conditional Statement 

## Type of Condition Statement 

#####  if Condition

```python
a = 10
if a == 10:
    print('Equile')
```

##### if-else Condition 

```python
a = 30
if a == 30:
    print("Equile")
else:
    print("Not equile")
```

##### if-elif-else Condition 

```python
money = int(input("Give me money :"))

if money <= 10:
    print(f"I will get icecream. I have {money} taka")
elif money <= 20 and 20 > 10:
    print(f"I will get chocober. I have {money} taka")
else:
    print(f"I will get cone. I have {money} taka")
    
```

## Conditional Practice Question 

#### Gender Greeting Massage

```python
gender = input("Give your gender (Male or Female) :").strip() .lower()
name = input("Your name :").strip()
if gender == "male":
    print(f"Good Morning Mr.{name}")
elif gender == "female":
    print(f"Good morning Mst.{name}")
else:
    print(f"Invalid Input {gender}. Please give us valid gender")
```

#### Odd and Even Checker

```python
number = int(input('Give your number :'))

if number % 2 == 0:
    print(f"{number} is even number")
else :
    print(f'{number} is odd number') 
```

#### Age Check

```python
age = int(input("Give your age"))
name = input('Give your name')

if age >= 18:
    print(f'Hello {name} you are a valid voter')
else:
    print(f'Hello {name} you are too young for vote. You vaild for Vote around {18 - age} years latter')
```

#### Leap year Dediction 

```python
year = int(input('Give year :'))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is leap year")
else:
    print(f"{year} year is not leap year")
```
