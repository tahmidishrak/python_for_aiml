String Slicing Example :

```python
name = "Tahmid Ishrak Tanim"
print(name[0:6:])
print(name[7:13:])
print(name[14::])
print(name[::])
```

Type Conversion Examples :

```python
# Intiger number to String
num = 20
print(num)
print(type(num))
num = str(num)
print(num)
print(type(num))
```

```python
# Normal Number to Complex Number
num = 20
print(num)
print(type(num))
complex_num = complex(num)
print(complex_num)
print(type(complex_num))
```

Implicit Examples

```python
num = 12
print(num)
print(type(num))
divide = (num/2)
print(divide)
print(type(divide))
```
