# OOPS / Object-Oriented Programming in Python

## Classe in OOPs

```python
# Syntax of class
class Car:
    brand = 'Toyota Nissan Skyline GTR' # Attributes

    def car(self): # Method
        print("BMW M3 GTR") 
    print("Hello this is a class and it initialise")
print(Car())
```

```python
# Accessing Attributes and Methods
class Car:
    brand = 'Toyota Nissan Skyline GTR . This is a Attribute' # Attributes

    def car(self): # Method
        print("BMW M3 GTR . This is a Method") 
    print("Hello this is a class and it initialize")
print(Car().brand)
Car().car()
```

## Obeject in OOPs

```python
# Syntax of Object
class Car:
    brand = 'Toyota Nissan Skyline GTR . This is a Attribute' # Attributes

    def car(self): # Method
        print("BMW M3 GTR . This is a Method") 
obj = Car() # object is we create a varibale and store the class and use this object multiple time
print(obj.brand)
obj.car()
```
