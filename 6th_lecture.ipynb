## List in Python

### List Power

#### Mutable

```python
# We can modify our list without create new one 
shopping_list = ["apples", "milk", "bread"]
print("Original list:", shopping_list)

shopping_list[1] = "almond milk" # Hare we can modify our list by using index value
print("After modification:", shopping_list)
```

#### Duplicates

```python
# The original list with repeated items
my_list = [1, 2, 3, 1, 2, 4, 5, 2]

seen = set()
duplicates = set()

for item in my_list:
    if item in seen:
        duplicates.add(item)
    else:
        seen.add(item)

# Convert back to a list to use the results
print(list(duplicates))  
# Output: [1, 2]
```

#### Ordered

```python
# 1. Create an ordered list
fruits = ["banana", "apple", "cherry"]
print("Original insertion order:", fruits)
# Output: ['banana', 'apple', 'cherry']

# 3. Access items by index (0-indexed)
print("First item:", fruits[0])   # Output: banana
print("Second item:", fruits[1])  # Output: apple
```

#### Heterogeneous

```python
# A heterogeneous list containing a string, integer, float, boolean, and another list
mixed_list = ["Alice", 42, 3.14, True, [1, 2], print()]

print(mixed_list)
# Output: ['Alice', 42, 3.14, True, [1, 2]]
```

### Set Operations 

#### Union

```python
a = {1,2,3,4,5,6,7,7,8,9,9}
b = {1,2,34,34,5,61,72,43,67,45,7}
print(a.union(b)) # Combine all sets
print(a | b)
```

####  Intersection

```python
a = {1,2,3,4,5,6,7,7,8,9,9}
b = {1,2,34,34,5,61,72,43,67,45,7}
print(a.intersection(b)) # Return only common element in the set
print(a & b)
```

#### Difference

```python
a = {1,2,3,4,5,6,7,7,8,9,9}
b = {1,2,34,34,5,61,72,43,67,45,7}
print(a.difference(b)) 
print(a - b) # Return element in the first set that are not have in the second set
```

#### Symmetric Difference

```python
a = {1,2,3,4,5,6,7,7,8,9,9}
b = {1,2,34,34,5,61,72,43,67,45,7}
print(a.symmetric_difference(b))
print(a ^ b)
```

## Dictionary

#### Dictionary Syntex 

```python
dic = {"name" : "tahmid",
      "age" : 15,
      "class" : 9}
print(dic["age"])
```

```python
# CRUD
dic = {"name" : "tahmid",
      "age" : 15,
      "class" : 9}
dic["profession"] = 'engineer' # Create a new key and value
dic["name"] = "ishrak" # Updateing the value
del dic["class"]
print(dic)
```

### Dictrionary Traversing

```python
dic = {"name" : "tahmid",
      "age" : 15,
      "class" : 9}
for i in dic:
    print(i)
print("------------")
for i in dic:
    print(dic[i])
print("------------")
for i in dic.values():
    print(i)
print("------------")
for i in dic:
    print(i ,":" ,dic[i])
```

### Difference Between Deep Copy and Shallow Copy

```python
# Normal Copy without Nasted value
dic = {"name" : "tahmid",
      "age" : 15,
      "class" : 9}
d = dic.copy()
print(d)
del d["class"]
print(d)
print(dic)
```

```python
# Copy with Nasted value
dic = {"name" : "tahmid",
      "mark": [100,80,60]}
d = dic.copy()
print(f"Copied Dictionary before change {d}")
d['mark'][0] = 90
print(f"Copied Dictionary after change {d}")
print(f"Original Dictionary {dic}")
```

```python
# Deep Copy With nasted value
import copy
dic = {"name" : "tahmid",
      "mark": [100,80,60]}
d = copy.deepcopy(dic)
print(f"Copied Dictionary before change {d}")
d['mark'][0] = 90
print(f"Copied Dictionary after change {d}")
print(f"Original Dictionary {dic}")
```

### Dictionary Question

```python
# Write a Python script to merge two Python dictionaries
a = {"name" : "tahmid",
    "age": 15}
b = {'city': 'dhaka',
     'kilo': 1600}
for i in b:
    a[i] = b[i]
print(a)
```

```python
# Write a Python program to sum all the values in a dictionary
a = {1: 100,
    2: 90,
    4: 43}
sum_dict = 0
for i in a.values():
    sum_dict += i
print(sum_dict)
```

```python
# Count the frequency of each element
a = [1,1,1,1,1,1,1,4,4,4,44,444,444,33,333,333,3,2,2,22,22,2,2,2,2,4,4,4,4,4,4,4,44,4,4,444,4,44,3,3,3,33,33]
b = {}
for i in a:
    if i in b.keys():
        b[i] += 1
    else:
        b[i] = 1
print(b)
```

```python
# Write a Python program to combine two dictionary by adding values for common keys.
a = {10: 100,50: 300, 32: 99}
b = {10: 211, 40: 988, 32: 900}
for i in b:
    if i in a.keys():
        a[i] += b[i]
    else:
        a[i] = b[i]
print(a)
```
