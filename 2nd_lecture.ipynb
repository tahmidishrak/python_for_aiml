# Operators

## Arithmatic Operators Examples :

### Different Bitween Division and Floor Division

```python
a = 50
b = 5
division = a / b
print(division)
print(type(division))

print("---------------")

floor_division = a // b
print(floor_division)
print(type(floor_division))
```

### Compund Assignment Operators and Reassignment Examples :

```python
num = 20
print(num)

print("--------")
num += 20 #num = num + 20 #also that's called reassingment
print(f"first ressignment :{num}")

print("--------")
num += 40
print(f"second ressignment :{num}")

print("--------")
num += 80
print(f"third ressignment :{num}")
```

Comparison Operators

## Comparison operator 

### Comparison operator provide always Boolean give answer True or False

```python
num = 3
num2 = 4
com = (num != num2)
print(type(com))
print(com)

print("---------------")

num = 3
num2 = 4
com = (num >= num2)
print(type(com))
print(com)
```

### Strings will be comparing the Ascii values of string.

```python
latter1 = "T"
print(ord(latter1[0]))
latter2 = "I"
print(ord(latter2[0]))

compair = (latter1 > latter2)
print(type(compair))
print(compair)
```
