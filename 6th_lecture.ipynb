## List in Python

### List Power

#### Mutable

```python
# We can modify our list without create new one 
shopping_list = ["apples", "milk", "bread"]
print("Original list:", shopping_list)

shopping_list[1] = "almond milk" # Hare we can modify our list by using index value
print("After modification:", shopping_list)
```

#### Duplicates

```python
# The original list with repeated items
my_list = [1, 2, 3, 1, 2, 4, 5, 2]

seen = set()
duplicates = set()

for item in my_list:
    if item in seen:
        duplicates.add(item)
    else:
        seen.add(item)

# Convert back to a list to use the results
print(list(duplicates))  
# Output: [1, 2]
```

#### Ordered

```python
# 1. Create an ordered list
fruits = ["banana", "apple", "cherry"]
print("Original insertion order:", fruits)
# Output: ['banana', 'apple', 'cherry']

# 3. Access items by index (0-indexed)
print("First item:", fruits[0])   # Output: banana
print("Second item:", fruits[1])  # Output: apple
```

#### Heterogeneous

```python
# A heterogeneous list containing a string, integer, float, boolean, and another list
mixed_list = ["Alice", 42, 3.14, True, [1, 2], print()]

print(mixed_list)
# Output: ['Alice', 42, 3.14, True, [1, 2]]
```

### List Traversing and Methods 

```python
a = [2, 5, 6, 7, 90]
for i in range(len(a)):
    print(i,"-",a[i])
```

#### List Method

```python
# Append 
a = [2, 5, 6, 7, 90]
a.append(100)
print(a)
```

```python
# Insert
a = [2, 5, 6, 7, 90]
a.insert(2,50)
print(a)
```

```python
# Extend
a = [2, 5, 6, 7, 90]
a.extend([8,80,76,77,998])
print(a)
```

```python
# Remove
a = [2, 5, 6, 7, 90]
a.remove(5)
print(a)
```

```python
# Pop
a = [2, 5, 6, 7, 90]
poped_item = a.pop(2)
print(poped_item)
```

```python
# Index
a = [2, 5, 6, 7, 90]
index = a.index(90)
print(index)
```

```python
# Sort
a = [2, 9, 8, 7, 90]
a.sort()
print(a)
```

```python
# Revers
a = [2, 5, 6, 7, 90]
a.reverse()
print(a)
```

```python
# Copy
a = [2, 5, 6, 7, 90]
new_num = a.copy()
print(new_num)
```

```python
# Clear
a = [2, 5, 6, 7, 90]
a.clear()
print(a)
```

### List Question 

```python
# Print positive and negative elements of an List?
a = [-90,-100,-39,89,43,21,23,-99]
print("Positive Element are :")
for i in a:
    if i >= 0:
        print(i)
print("Nagetive Element are :")
for i in a:
    if i <= 0:
        print(i)
        
```

```python
# Mean of List elements?
a = [1,4,5,6,7,4,9,50,9]
sum_of_a = 0
for i in a:
    sum_of_a += i
    mean = sum_of_a/len(a)
print(mean)
```

```python
# Find the greatest element and print its index too?
a = [1,4,5,6,7,4,9,50,9]
largest = a[0]
index = 0
for i in range(len(a)):
    if a[i] > largest:
        largest = a[i]
        index = i
print(f"Your largest number is {largest} and index value is {index}")
    
```
