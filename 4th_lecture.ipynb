# Loops in Python

#### Loops with Number. Using Range Function

```python
num = range(1, 11, 1) 
""" important note: If you want the loop to run until 10, you have to write 11 in the code,
because the end value in range is always excluded. """
for i in num:
    print(i)

print("---------------------")

#We can write range in different way
for i in range(1, 11, 2): # This way is fast and easy
    print(i)
```

##### Number Reverse print. And print specific number to specific number

```python
# specific number to specific number
for i in range(20, 25):
    print(i)

print("----------------")

# Number Reverse print
for i in range(5, 0, -1):
    print(i)

# Nagetive number to Negetive number
print("----------------")
for i in range(-4, -9, -1):
    print(i)
```

```python
# Multiply Table of 23

for i in range (23, 231, 23):
    print(i)
```

##### Create a Multiply Table using form user input

```python
num = int(input("Which Table you want? Give this number :"))

for i in range(num , (num*10)+1, num):
    print(i)
```

#### For Loop with String 

- Using Index Value

```python
name = "Tahmid"
for i in range(len(name)): # Here we use len() function for take out the variable length
    print(name[i]) 
# Note:
# 'i' holds the position number (Index: 0, 1, 2...).
# 'name[i]' fetches the actual character at that index ('T', 'a', 'h'...).
```

- Iterating directly over the string

```python
name = "Tahmid"
for i in name:
    print(i) 
```

#### Break Continue Else

- Break Statement 

```python
for i in range(1, 11, 1):
    if i == 9: # Here we use if condition for check (i == 9).
        break # And if (i == 9) break stop the loop 
    else:
        print(i) # And here else statement work is when i != 9 then print other like (1, 2, 3...)
```

- Continue Statement

```python
for i in range(1, 11, 1):
    if i == 9 or i == 8: # Here we use if condition for check (i == 9).
        continue # And if (i == 9) skip this iteration and jump to the next iteration
    else:
        print(i) # And here else statement work is when i != 9 then print other like (1, 2, 3...)
```

- Else statement

```python
# here the break statement is executed for this reasone else is not executed 
for i in range(1, 11, 1):
    if i == 9:
        print("Break statement is executed")
        break
    else:
        print(i)
else:
    print("Break statement is not executed")

print("-----------------------------")

# here the break statement is not executed for this reasone else is executed
for i in range(1, 11, 1):
    if i == 19:
        print("Break statement is executed")
        break
    else:
        print(i)
else:
    print("Break statement is not executed")
```

- Seventh Question

```python
# Print the sum of all even & odd numbers

n = int(input("Give your number :"))
sum_even = 0
sum_odd = 0
for i in range(1, n+1):
    if i % 2 == 0:
        sum_even += i
    else:
        sum_odd += i

print(f"Sum of even number is : {sum_even}")
print(f"Sum of odd number is : {sum_odd}")
```
