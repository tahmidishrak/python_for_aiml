#### Loops with Number. Using Range Function

```python
num = range(1, 11, 1) 
""" important note: If you want the loop to run until 10, you have to write 11 in the code,
because the end value in range is always excluded. """
for i in num:
    print(i)

print("---------------------")

#We can write range in different way
for i in range(1, 11, 2): # This way is fast and easy
    print(i)
```

#### Break Continue Else

- Break Statement 

```python
for i in range(1, 11, 1):
    if i == 9: # Here we use if condition for check (i == 9).
        break # And if (i == 9) break stop the loop 
    else:
        print(i) # And here else statement work is when i != 9 then print other like (1, 2, 3...)
```

- Continue Statement

```python
for i in range(1, 11, 1):
    if i == 9 or i == 8: # Here we use if condition for check (i == 9).
        continue # And if (i == 9) skip this iteration and jump to the next iteration
    else:
        print(i) # And here else statement work is when i != 9 then print other like (1, 2, 3...)
```

- Else statement

```python
# here the break statement is executed for this reasone else is not executed 
for i in range(1, 11, 1):
    if i == 9:
        print("Break statement is executed")
        break
    else:
        print(i)
else:
    print("Break statement is not executed")

print("-----------------------------")

# here the break statement is not executed for this reasone else is executed
for i in range(1, 11, 1):
    if i == 19:
        print("Break statement is executed")
        break
    else:
        print(i)
else:
    print("Break statement is not executed")
```

### While loop Question

- First Question 

```python
# Separate each digit of a number and print it on the new line

a = int(input("give your number :"))
while a > 0:
    print(a % 10)
    a //= 10
```

- Second Question

```python
# Accept a number and print its reverse

a = int(input("give your number :"))

revers_digit = 0
while a > 0:
    last_digit = a % 10
    revers_digit = (revers_digit * 10) + last_digit
    a //= 10
print(f"Reversed number is: {revers_digit}")
```

- Third Question

```python
# Accept a number and check if it is a pallindromic number (If number and its reverse are equal?

a = int(input("give your number :"))
original_num = a
revers_digit = 0
while a > 0:
    last_digit = a % 10
    revers_digit = (revers_digit * 10) + last_digit
    a //= 10

if original_num == revers_digit:
    print(f"Your Number is pallandrome : {original_num}")
else:
    print(f"Your Number is not pallandrome : {original_num}")
        
```

#### Create a random number guessing game with python. 

```python
import random 
num = random.randint(1,100)
tries = 0
max_tries = random.randint(1,15)
print(f"You have {max_tries} tries to find the right number")
while tries < max_tries:
    guess = int(input("give your number between 1 to 100 :"))
    tries += 1
    if num == guess:
        print(f"Your are right.You guessed the number is {tries} tries")
        break
    else:
        if tries == max_tries:
            print("You lose")
        elif num < guess:
            print("OOPs! Go a little lower")
        elif num > guess:
            print("OOPs! Go a little higher")
            
if num != guess:
    print(f"You reached your limit. The number was {num}. You used {tries} tries.")
```
