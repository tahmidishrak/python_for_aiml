# OOPS / Object-Oriented Programming in Python

## Classe in OOPs

```python
# Syntax of class
class Car:
    brand = 'Toyota Nissan Skyline GTR' # Attributes

    def car(self): # Method
        print("BMW M3 GTR") 
    print("Hello this is a class and it initialise")
print(Car())
```

```python
# Accessing Attributes and Methods
class Car:
    brand = 'Toyota Nissan Skyline GTR . This is a Attribute' # Attributes

    def car(self): # Method
        print("BMW M3 GTR . This is a Method") 
    print("Hello this is a class and it initialize")
print(Car().brand)
Car().car()
```

## Obeject in OOPs

```python
# Syntax of Object
class Car:
    brand = 'Toyota Nissan Skyline GTR . This is a Attribute' # Attributes

    def car(self): # Method
        print("BMW M3 GTR . This is a Method") 
obj = Car() # object is we create a varibale and store the class and use this object multiple time
print(obj.brand)
obj.car()
```

### Constructor in OOPs

```python
class Robot:
    # The constructor method
    def __init__(self, name, color):
        self.name = name    # Instance attribute
        self.color = color  # Instance attribute

    def introduce_self(self):
        print(f"Hello, my name is {self.name} and I am {self.color}.")

# Creating objects (This automatically calls the __init__ constructor)
robot1 = Robot("Robbie", "red")
robot2 = Robot("Sparky", "blue")

# Accessing attributes and methods
robot1.introduce_self()  
robot2.introduce_self()  
```

## Attributes and Methods 

```python
class Car:
    # Class attribute (All cars share this trait)
    wheels = 4 

    def __init__(self, color, brand):
        # Instance attributes (Unique to each car)
        self.color = color
        self.brand = brand

# Creating two separate instances (objects)
car1 = Car("Red", "Toyota")
car2 = Car("Blue", "Ford")

# 1. Reading Attributes
print(car1.brand)   # Output: Toyota (Instance attribute)
print(car2.brand)   # Output: Ford   (Instance attribute)
print(car1.wheels)  # Output: 4      (Shared class attribute)
print(car2.wheels)  # Output: 4      (Shared class attribute)

# 2. Modifying an Instance Attribute
car1.color = "Green"
print(car1.color)   # Output: Green  (Only car1 changes)
print(car2.color)   # Output: Blue   (car2 is completely unaffected)

# 3. Modifying the Class Attribute
Car.wheels = 6
print(car1.wheels)  # Output: 6      (Changes for car1)
print(car2.wheels)  # Output: 6      (Changes for car2 because they share it)
```

```python
class BankAccount:
    # Class Attribute (shared across all instances)
    bank_name = "City Bank"

    def __init__(self, owner, balance):
        # Instance Attributes (unique to each instance)
        self.owner = owner
        self.balance = balance

    # 1. Instance Method
    # Purpose: Operates on data belonging to a specific object.
    # Feature: Takes 'self' as the required first parameter.
    def deposit(self, amount):
        self.balance += amount
        print(f"{self.owner} deposited {amount}. New Balance: {self.balance}")

    # 2. Class Method
    # Purpose: Operates on class-level attributes that affect all instances.
    # Feature: Uses the '@classmethod' decorator and takes 'cls' as the first parameter.
    @classmethod
    def change_bank_name(cls, new_name):
        cls.bank_name = new_name
        print(f"Bank name changed to {cls.bank_name}")

    # 3. Static Method
    # Purpose: Acts as a utility/helper function isolated inside the class.
    # Feature: Uses the '@staticmethod' decorator; does not receive 'self' or 'cls'.
    @staticmethod
    def is_valid_account_number(acc_num):
        # Validates if the account number consists of exactly 10 digits
        return len(acc_num) == 10 and acc_num.isdigit()


# ==================== TESTING THE CODE ====================

# 1. Calling Instance Method
acc1 = BankAccount("Tahmid", 5000)
acc1.deposit(1500)  # Output: Tahmid deposited 1500. New Balance: 6500

# 2. Calling Class Method (directly using the class name)
BankAccount.change_bank_name("DBBL")  # Output: Bank name changed to DBBL

# 3. Calling Static Method (directly using the class name)
print(
    BankAccount.is_valid_account_number("1234567890")
)  # Output: True (10 digits)
```
