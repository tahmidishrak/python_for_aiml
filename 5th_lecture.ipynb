# Function in Python 

##### Difference between Paremeter and Argument

```python
#Parameter is a varibale and this store argumnet 
def greet(name):  # "name" is the PARAMETER
    print("Hello " + name)

# Argumnet is actual data or value that provid user when fucntion called
greet(input("Give your name :"))  # "Alice" is the ARGUMENT

# that's means parameter is varibale and argument is data
```

## Type of Argument 

### Positional Argument

```python
def add(a, b):
    print(f"Your sum is {a + b}")

add(3,4)
```

### Keyword Argument

```python
def hello(name, age):
    print(f"Hello {name}, Your age is {age}")
hello(name = 'Tahmid', age = 15)
```

### Default Argument

```python
def my_function(name = "friend"):
  print("Hello", name)

my_function("Tahmid")
my_function()
print(type(my_function()))
```

## Function Question

#### Create function for check Palindrome 

```python
def palindrome(name):
    rev = ""
    if rev == name[::-1]: # Here we use slicing Method for reversing String 
        print(f"{name} is Pallindrome")
    else:
        print(f"{name} is not Pallindrome")

palindrome("Tahmid")
```
